```python
df["Date"] = pd.to_datetime(
    df["Year"].astype(str) + "-" + df["Month"],
    format="%Y-%B",
    errors="coerce"
)

print(df[["Year", "Month", "Date"]].head())
print(df["Date"].dtype)
```

```python
sns.boxplot(x=df["Revenue"])

plt.title("Revenue Outliers")
plt.show()
```

```python
q1 = df["Revenue"].quantile(0.25)
q3 = df["Revenue"].quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

df = df[
    (df["Revenue"] >= lower_bound) &
    (df["Revenue"] <= upper_bound)
].copy()

print("Outliers Removed")
print("Updated Dataset Shape:", df.shape)
```

```python
print(df["Category"].value_counts())

plt.figure(figsize=(10, 5))

sns.countplot(
    data=df,
    y="Category",
    order=df["Category"].value_counts().index
)

plt.title("Category Distribution")
plt.xlabel("Count")
plt.ylabel("Category")
plt.show()
```

```python
category_sales = (
    df.groupby("Category")["Revenue"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(category_sales)

category_sales.plot(kind="bar", figsize=(10, 5))

plt.title("Top Categories by Revenue")
plt.xlabel("Category")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)
plt.show()
```

```python
top_categories = df["Category"].value_counts().head(5).index

subset = df[df["Category"].isin(top_categories)]

plt.figure(figsize=(12, 6))

sns.boxplot(
    data=subset,
    x="Revenue",
    y="Category",
    hue="Region"
)

plt.title("Revenue by Category and Region")
plt.xlabel("Revenue")
plt.ylabel("Category")
plt.show()
```
